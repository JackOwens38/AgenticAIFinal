```python
import os, shutil, zipfile
from google.colab import userdata, drive

# Your OpenAI key lives in Colab Secrets (the key icon). LangChain looks for it in this environment variable.
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
drive.mount("/content/drive")      # gives this notebook access to your Google Drive

# The LangChain pieces we use today:
from langchain_openai import ChatOpenAI, OpenAIEmbeddings          # OpenAI's chat model and embedding models
from langchain_chroma import Chroma                                # Chroma, our vector database
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage   # the message types in a chat
from langchain_core.tools import tool                              # turns a Python function into a tool
from langchain.agents import create_agent                          # builds the agent loop for us (Step 6)
```

```python
# temperature=0 means the model picks its most likely words, so everyone's output is close to the same.
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)                 # writes the replies
```

```python
# Each llm.invoke() is a brand-new conversation: the model sees ONLY the text we pass in that call.
print("Call 1:", llm.invoke("What did Zhang Huan use to create his 1965 Last Supper artwork?").content)
print("Call 2:", llm.invoke("Who created Wall of Light Black in 1998?").content)
print("Call 3:", llm.invoke("What colors are used in the painting Spring by Natalia Goncharova?").content)
print("Call 4:", llm.invoke("What artifacts are used in the White Rit sculture created by Vanessa German?").content)
print("Call 5:", llm.invoke("In what year did Francesco de Mura create Bacchus and Ceres?").content)
print("Call 6:", llm.invoke("Who gifted the 1786 painting Marguerite Baudard de Saint James to the Raclin Murphy?").content)
print("Call 7:", llm.invoke("Why is The Artist’s Studio, a Sketch by Jean-Léon Gérôme an important piece in the Raclin Murphy Museum?").content)
print("Call 8:", llm.invoke("Have any pieces of art been donated as past of the Bernard Norling and Mary T. Norling Endowment for 18th- and 19th-Century Sculpture?").content)
```
